```python
# unzip cards.zip

import zipfile
with zipfile.ZipFile('cards.zip', 'r') as zip_ref:
    zip_ref.extractall('cards')
```

Розархівуємо файл, закидуємо необхідну папку з даними в робочу директорію. Видаляємо зайве

```python
# move directory
import shutil
shutil.copytree('cards/cards', 'cards', dirs_exist_ok=True)
```

```python
shutil.rmtree('cards/cards')
```
